# Ejercicio 1

Resolver la siguiente ecuación diferencial

\[
\frac{dy}{dx}=\sin(5x)
\]

---

```python
x = sp.Symbol('x')
y = sp.Function('y')

f = sp.sin(5*x)

ecuacion = sp.Eq(y(x).diff(x),f)

display(ecuacion)

solucion = sp.dsolve(y(x).diff(x)-f)

display(solucion)
```

## Procedimiento

Separamos variables

\[
dy=\sin(5x)\,dx
\]

Integramos ambos lados

\[
\int dy=\int\sin(5x)\,dx
\]

Sabemos que

\[
\int\sin(ax)\,dx=-\frac1a\cos(ax)
\]

Entonces

\[
y=-\frac15\cos(5x)+C
\]

### Resultado

\[
\boxed{y=-\frac15\cos(5x)+C}
\]

\]

Integramos

\[
-\frac12e^{-2y}=\frac13e^{3x}+C
\]

Despejando

\[
\boxed{y=-\frac12\ln\left(C-\frac23e^{3x}\right)}

# Ejercicio 8

Resolver

\[
e^y\frac{dy}{dx}=e^x+e^{2x-y}
\]

---

```python
x = sp.Symbol('x')
y = sp.Function('y')

f = (sp.exp(x)+sp.exp(2*x-y(x)))/sp.exp(y(x))

ecuacion = sp.Eq(y(x).diff(x), f)

display(ecuacion)

try:
    display(sp.dsolve(y(x).diff(x)-f))
except:
    print("SymPy no encuentra una solución explícita.")
```

## Procedimiento

Observamos que

\[
e^y\frac{dy}{dx}=\frac{d}{dx}(e^y)
\]

Entonces

\[
d(e^y)=(e^x+e^{2x})dx
\]

Integramos

\[
e^y=e^x+\frac12e^{2x}+C
\]

Finalmente

\[
\boxed{y=\ln\left(e^x+\frac12e^{2x}+C\right)}
\]

# Ejercicio 19

Resolver

\[
\frac{dy}{dx}
=
\frac{xy+3x-y-3}
{5y-2x-8}
\]

---

```python
x = sp.Symbol('x')

y = sp.Function('y')

f = (x*y(x)+3*x-y(x)-3)/(5*y(x)-2*x-8)

ecuacion = sp.Eq(y(x).diff(x),f)

display(ecuacion)

try:
    display(sp.dsolve(y(x).diff(x)-f))
except:
    print("SymPy no puede resolver esta ecuación.")
```

## Procedimiento

La ecuación dada

\[
\frac{dy}{dx}
=
\frac{xy+3x-y-3}
{5y-2x-8}
\]

no corresponde al método de separación de variables.

Al factorizar el numerador

\[
xy+3x-y-3
=
(x-1)(y+3)
\]

el denominador no permite separar completamente las variables.

**Conclusión**

Esta ecuación probablemente contiene un error de impresión en el problemario original.

# Ejercicio 20

Resolver

\[
\frac{dy}{dx}
=
\frac{xy+2x-y-2}
{xy-3y+x-3}
\]

---

```python
x = sp.Symbol('x')

y = sp.Function('y')

f = (x*y(x)+2*x-y(x)-2)/(x*y(x)-3*y(x)+x-3)

ecuacion = sp.Eq(y(x).diff(x),f)

display(ecuacion)

display(sp.dsolve(y(x).diff(x)-f))
```

## Procedimiento

Factorizando

\[
xy+2x-y-2
=
(x-1)(y+2)
\]

\[
xy-3y+x-3
=
(x-3)(y+1)
\]

Separando variables

\[
\frac{y+1}{y+2}dy
=
\frac{x-1}{x-3}dx
\]

Integramos

\[
y-\ln|y+2|
=
x+2\ln|x-3|+C
\]

### Resultado

\[
\boxed{
y-\ln|y+2|
=
x+2\ln|x-3|+C
}
\]
