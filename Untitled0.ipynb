```python
# Importar bibliotecas

import sympy as sp

sp.init_printing()

t, s = sp.symbols('t s', positive=True)
```

# Ejercicio 1

Calcular

\[
\mathcal{L}\{te^{4t}\}
\]

```python
f = t*sp.exp(4*t)

L = sp.laplace_transform(f,t,s,noconds=True)

print("Transformada:")
display(L)
```

# Ejercicio 2

\[
\mathcal{L}\{t^2e^{-2t}\}
\]

```python
f = t**2*sp.exp(-2*t)

L = sp.laplace_transform(f,t,s,noconds=True)

display(L)
```

# Ejercicio 3

\[
\mathcal{L}\{e^{-t}\sin(t)\}
\]

```python
f = sp.exp(-t)*sp.sin(t)

display(sp.laplace_transform(f,t,s,noconds=True))
```

# Ejercicio 4

\[
\mathcal{L}\{e^t\cos(t)\}
\]

```python
f = sp.exp(t)*sp.cos(t)

display(sp.laplace_transform(f,t,s,noconds=True))
```

# Ejercicio 5

\[
\mathcal{L}\{t\cos(t)\}
\]

```python
f = t*sp.cos(t)

display(sp.laplace_transform(f,t,s,noconds=True))
```

# Ejercicio 6

\[
\mathcal{L}\{t\sin(t)\}
\]

```python
f = t*sp.sin(t)

display(sp.laplace_transform(f,t,s,noconds=True))
```

# Ejercicio 7

\[
\mathcal{L}\{\cos^2(t)\}

# Transformadas Inversas

```python
F = 1/s**4

display(sp.inverse_laplace_transform(F,s,t))
```

```python
F = 1/s**2 - 1/s + 1/(s-2)

display(sp.inverse_laplace_transform(F,s,t))
```

```python
F = (2*s-6)/(s**2+9)

display(sp.inverse_laplace_transform(F,s,t))
```

```python
F = 4/s + 6/s**5 - 1/(s+8)

display(sp.inverse_laplace_transform(F,s,t))
```

```python
F = 1/(5*s-2)

display(sp.inverse_laplace_transform(F,s,t))
```

```python
F = 1/(4*s**2+1)

display(sp.inverse_laplace_transform(F,s,t))
```

```python
F = (2*s-6)/(s**2+9)

display(sp.inverse_laplace_transform(F,s,t))
```

```python
F = (s+1)/(s**2+2)

display(sp.inverse_laplace_transform(F,s,t))
```

```python
F = (s+1)/(s**2-4*s)

display(sp.inverse_laplace_transform(F,s,t))
```

```python
F = 1/(s**2+s-20)

display(sp.inverse_laplace_transform(F,s,t))
```
